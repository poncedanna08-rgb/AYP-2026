Manejo de Datos Numericos


```python
import time
lista = list(range (10000000))
inicio = time.time()
[valor * 2 for valor in lista]
print ("Tiempo de ejecucion con listas: ", time.time()-inicio)
```

```python
import numpy as np
arreglo = np.arange(10000000)
inicio = time.time()
arreglo * 2
print ("Tiempo de ejecucion con listas: ", time.time()-inicio)
```

Dimensiones y Formas de los Arreglos

```python
print("El numero de dimensiones del arreglo es: ", arreglo.ndim)
print ("La forma del arreglo es: ", arreglo.shape)
print ("El total de elementos en el arreglo es: ", arreglo.size)
print ("Los tipos de datos en el arreglo son: ", arreglo.dtype)
```
