```python
df = df.drop(columns=["Unnamed: 0"])
```

```python
df["Age of Mother"] = df["Age of Mother"].fillna(
    df["Age of Mother"].median()
)

df["Age of Father"] = df["Age of Father"].fillna(
    df["Age of Father"].median()
)

df["Age at First Diagnosis"] = df["Age at First Diagnosis"].fillna(
    df["Age at First Diagnosis"].median()
)
```

```python
df["Diagnosis Group"] = pd.cut(
    df["Age at First Diagnosis"],
    bins=[0,5,12,18,100],
    labels=[
        "Early Childhood",
        "Childhood",
        "Adolescence",
        "Adult"
    ]
)
```

```python
df["Average Parent Age"] = (
    df["Age of Mother"] +
    df["Age of Father"]
) / 2
```

```python
df.columns
```

```python
symptoms = [
    'Café au lait (CLS)',
    'Axillary Freckles',
    'Inguinal Freckles',
    'Lisch Nodules',
    'Dermal Neurofibromins',
    'Plexiform Neurofibromins',
    'Optic Glioma',
    'Skeletal Dysplasia',
    'Learning Disability',
    'Hypertension',
    'Astrocytoma',
    'Hamartoma',
    'Scoliosis'
]

df["Total Symptoms"] = df[symptoms].sum(axis=1)
```

```python
df["Risk Level"] = pd.cut(
    df["Total Symptoms"],
    bins=[0,3,6,20],
    labels=[
        "Low",
        "Medium",
        "High"
    ]
)
```

```python
df.to_excel(
    "Neurofibromatosis_Limpia.xlsx",
    index=False
)
```

```python
plt.figure(figsize=(8,5))
sns.histplot(
    df["Age at First Diagnosis"],
    bins=15,
    kde=True
)
plt.title("Distribution of Age at First Diagnosis")
plt.show()
```

```python
plt.figure(figsize=(6,5))
sns.countplot(
    x="Tumour Case",
    data=df
)
plt.title("Tumour Cases")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(
    x="Tumour Case",
    y="Total Symptoms",
    data=df
)
plt.title("Total Symptoms vs Tumour Presence")
plt.show()
```

```python
plt.figure(figsize=(7,5))
sns.countplot(
    x="Risk Level",
    data=df
)
plt.title("Risk Level Distribution")
plt.show()
```

```python
plt.figure(figsize=(12,8))

corr = df.select_dtypes(include=np.number).corr()

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Matrix")
plt.show()
```

```python
plt.figure(figsize=(7,7))

df["Tumour Case"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%",
    labels=["No Tumour", "Tumour"]
)

plt.title("Distribution of Tumour Cases")
plt.ylabel("")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.barplot(
    data=df,
    x="Tumour Case",
    y="Total Symptoms",
    estimator=np.mean
)

plt.title("Average Number of Symptoms by Tumour Presence")
plt.xlabel("Tumour Case")
plt.ylabel("Average Symptoms")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    x="Diagnosis Group",
    hue="Tumour Case",
    data=df
)

plt.title("Tumour Cases by Diagnosis Group")
plt.show()
```
